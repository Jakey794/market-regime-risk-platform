# 03 Feature Engineering

Leakage-safe regime feature construction using package APIs only.
Trailing windows, chronological splits, and train-only scaling.


````python
from pathlib import Path

from mrrp.data.cache import load_parquet
from mrrp.features import EXPECTED_REGIME_FEATURE_COLUMNS
from mrrp.portfolio.config import load_portfolio_config
````

````python
prices_path = Path("data/processed/adjusted_close.parquet")
if not prices_path.exists():
    prices_path = Path("data/sample/synthetic_prices.parquet")
prices = load_parquet(prices_path)
portfolio = load_portfolio_config("configs/sample_portfolio.yaml")
print("Prices:", prices.shape, "as-of", prices.index.max().date())
print("Holdings:", list(portfolio.holdings.index))
````
